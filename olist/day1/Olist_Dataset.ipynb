```python
oord_df["review_comment_title"]=oord_df["review_comment_title"].fillna("No comment",inplace=True)
oord_df["review_comment_message"]=oord_df["review_comment_message"].fillna("No comment",inplace=True)
oord_df.isnull().sum()
```

#cleaning on orders dataset



```python
ood_df.head()
```

```python
ood_df["order_purchase_timestamp"]=pd.to_datetime(ood_df["order_purchase_timestamp"])
ood_df["order_approved_at"]=pd.to_datetime(ood_df["order_approved_at"])
ood_df["order_delivered_carrier_date"]=pd.to_datetime(ood_df["order_delivered_carrier_date"])
ood_df["order_delivered_customer_date"]=pd.to_datetime(ood_df["order_delivered_customer_date"])
ood_df["order_estimated_delivery_date"]=pd.to_datetime(ood_df["order_estimated_delivery_date"])
ood_df.isnull().sum()
```

counts the value for all the shipped and various statuses

```python
null_status = ood_df[ood_df['order_delivered_customer_date'].isna()]['order_status'].value_counts()
print(null_status)
```

creats a new row consiting what status our order is on

```python
def order_completion_status(row):
    if row['order_status'] == 'delivered':
        return 'Delivered'
    elif row['order_status'] in ['shipped', 'processing', 'approved', 'invoiced']:
        return 'Stuck/In Progress'
    elif row['order_status'] in ['cancelled', 'unavailable']:
        return 'Cancelled'
    else:
        return 'Other'
ood_df['completion_status'] = ood_df.apply(order_completion_status, axis=1)
print(ood_df['completion_status'].value_counts())
```
